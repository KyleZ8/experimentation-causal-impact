# Case 3 — Propensity Score Matching

Seeded synthetic session data with self-selection into directed search and a planted treatment effect.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name != 'experimentation-causal-impact':
    PROJECT_ROOT = PROJECT_ROOT.parents[1]
SRC_ROOT = PROJECT_ROOT / 'src'
if str(SRC_ROOT) not in sys.path:
    sys.path.insert(0, str(SRC_ROOT))

from causal_toolkit.psm import (
    TRUE_DIRECTED_SEARCH_EFFECT,
    estimate_psm_att,
    generate_case3_sessions,
    naive_difference,
)

sessions = generate_case3_sessions()
naive = naive_difference(sessions)
result, pairs, scored = estimate_psm_att(sessions)

print(f'naive_sales_lift=${naive:.2f}')
print(f'psm_att=${result.att:.2f}')
print(f'psm_ci=${result.ci_low:.2f},${result.ci_high:.2f}')
print(f'planted_effect=${TRUE_DIRECTED_SEARCH_EFFECT:.2f}')
print(f'mean_abs_smd={result.mean_abs_smd_before:.3f}->{result.mean_abs_smd_after:.3f}')
print(f'matched_pairs={result.n_pairs}')
```
